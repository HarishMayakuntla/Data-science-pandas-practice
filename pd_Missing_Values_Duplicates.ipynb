# Missing Values, Duplicates

## Missing Values

A missing value means some data is not available.

### example :

| Name   | Age | Salary |
| ------ | --: | -----: |
| Harish |  22 |  30000 |
| Rahul  | NaN |  40000 |
| Rohit  |  24 |    NaN |





NAN ---- > means not a number / missing value

```python
## create a DataFrame

import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Rohit", "Anil"],
    "Age": [22, None, 24, None],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

print(df)
```

## Finding missing values

### using isnull()  Or isna()

both are same

```python
df.isnull()
```

* True → missing value
* False → value exists

```python
df.isna()
```

## Count Missing Values

This is one of the most important commands
* counts no.of missing values

### using isnull().sum()

```python
df.isnull().sum()
```

Meaning:

* m Name → 0 missing
* Age → 2 missing
* Salary → 1 missing

## Find Rows Having Missing Values

This gives rows where at least one value is missing

```python
df[df.isnull().any(axis=1)]
```

## Find Columns Having Missing Values

```python
df.columns[df.isnull().any()]

```

```python
df.columns
```

# remove Missing Values

### use dropna()

* This removes rows containing missing values
* dropna() does not permanently change df by default

```python
df.dropna()
```

## Remove Rows Only When All Values Are Missing

```python
df.dropna(how='all')
```

## Remove Missing Values From a Specific Column

```python
df.dropna(subset=["Age"])  # Remove rowes where "Age" is missing
```

## Fill Missing Values

* Instead of deleting data, sometimes we want to replace missing values

 ###  use fillna()  ----> fill null values

```python
df.fillna(0)
df
```

## perticular column

```python

import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Rohit", "Anil"],
    "Age": [22, None, 24, None],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

df
```

```python
df['Age']
```

```python
df["Age"]=df['Age'].fillna(0)
```

```python
df['Age']
```

## Fill with mean()

```python
salary_mean=df['Salary'].mean()
df["Salary"]=df["Salary"].fillna(salary_mean)
```

```python
df
```

###  Fill with Median

```python
age_mean=df['Age'].median()
df['Age']=df['Age'].fillna(age_mean)
```

```python
df
```

### Fill With Mode

For categorical data

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Rohit", None],
    "Age": [22, None, 24, None],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Name"] = df["Name"].fillna(df["Name"].mode()[0])
```

```python
df
```

### Forward filling

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Rohit", "Anil"],
    "Age": [22, None, 24, None],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Salary"] = df["Salary"].ffill()
df
```

### backward Fill

```python
df["Age"] = df["Age"].bfill()
df
```

# Duplicates

A duplicate means the same record appears more than once

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Rohit", "Harish"],
    "Age": [22, 21, 24, 22],
    "Salary": [30000, 40000, 20000, 30000]
}

df = pd.DataFrame(data)

df
```

### find duplicates

-- Syntax:

        DataFrame.duplicated()

```python
df.duplicated() # it gives boolean , True means duplicate row
```

### Count of Duplicates

-- Syntax :

        DataFrame.duplicated().sum()

```python
df.duplicated().sum()
```

### Display Duplicate Rows

```python
df[df.duplicated()]
```

### Remove Duplicates Based on One Column

```python
df.drop_duplicates(subset="Name")
```

### Remove Duplicates

Syntax :  

      DataFrame.drop_duplicates()

```python
df.drop_duplicates()
```

### Keep Last Duplicate

```python
df.drop_duplicates(
    subset=["Name"],
    keep="last"
)
```

### Remove All Copies of Duplicates

```python
df.drop_duplicates(
    subset=['Name'],
    keep=False
)
```
