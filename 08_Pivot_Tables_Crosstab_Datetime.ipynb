## Pivot Table


A Pivot Table is used to summarize data.

It can calculate:

* Sum
* Mean
* Count
* Max
* Min

Instead of manually filtering data, pivot_table() creates a summary table.

```python
import pandas as pd

data = {
    "Department":["IT","HR","IT","HR","Sales","Sales"],
    "Employee":["A","B","C","D","E","F"],
    "Salary":[50000,40000,60000,45000,55000,52000]
}

df = pd.DataFrame(data)

df
```

```python
##  find mean

pd.pivot_table(
    df,
    values='Salary',
    index="Department",
    aggfunc='mean'


)
```

```python
# count of

pd.pivot_table(
    df,
    values='Salary',
    index="Department",
    aggfunc='count'
)
```

```python
#find sum

pd.pivot_table(

      df,
      values='Salary',
      index='Department',
      aggfunc='sum'

)
```

### Multiple Aggregations

```python
pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc=['sum','mean','std','count','min','max']
)
```

### Two - leval pivot table

```python
df['Gender']=['M','F','M','M','F','M']
df
```

```python
pd.pivot_table(

        df,
        values='Salary',
        index='Employee',
        columns='Gender',
        aggfunc=['sum','mean','count']
)
```

## missing values

```python
pd.pivot_table(

        df,
        values='Salary',
        index='Employee',
        columns='Gender',
        aggfunc=['sum','mean','count'],
        fill_value=0
)
```

# Crosstab

crosstab() counts how many times combinations of categories occur

```python
df
```

```python
pd.crosstab(
    df['Department'],
    df['Gender']
)
```

Grand percentage

```python
# it give  in percentage

pd.crosstab(
    df['Department'],
    df['Gender'],
    normalize=True
)
```

### Add toatals

```python
pd.crosstab(
    df['Department'],
    df['Gender'],
    margins=True
)
```

###========================= pivot  vs  crosstab =================

| Pivot Table                         | Crosstab                       |
| ----------------------------------- | ------------------------------ |
| Summarizes values                   | Counts category combinations   |
| Uses numeric values                 | Uses categorical data          |
| Supports `sum`, `mean`, `max`, etc. | Mainly frequency counts        |
| Requires `values` and `aggfunc`     | Works directly with categories |


# Datetime

Almost every real-world dataset has dates

```python
## covert string to Datatime

import pandas as pd

df = pd.DataFrame({

    "Date":["2026-08-01","2026-08-02","2026-08-03"]
})

df["Date"] = pd.to_datetime(df["Date"])

df
```

```python
df.dtypes
```

```python
## extract Year
df['Year'] = df['Date'].dt.year
df['Year']
```

```python
## extract month
df['Month'] = df['Date'].dt.month
df['Month']
```

```python
## extract day

df["Day"] = df["Date"].dt.day
df['Day']
```

```python
## day name
df["Date"].dt.day_name()
```

```python
# month name
df["Date"].dt.month_name()
```

```python
# weakday number
df["Date"].dt.weekday
```

## Date Difference

```python
df = pd.DataFrame({
    "Start":["2026-08-01"],
    "End":["2026-08-10"]
})

df["Start"] = pd.to_datetime(df["Start"])
df["End"] = pd.to_datetime(df["End"])

df["Days"] = df["End"] - df["Start"]
df
```

```python
## sort by dates

df.sort_values("Date")
```

```python
# Filter by Date

df[df["Date"] > "2026-08-01"]
```
