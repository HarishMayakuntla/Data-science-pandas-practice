# pands using Merge, Join, Concat

## Merge

Use merge() when two DataFrames have a common column

* Syntax:

           pd.merge(left, right, on="column_name")

```python
import pandas as pd

df1=pd.DataFrame({

        'ID':[1,2,3,4],
        'Name':['harish','ramesh','ganesh','mukesh']
})
```

```python
df1
```

```python
df2=pd.DataFrame({

    "ID": [1,2,3,4],
    "Marks": [80,90,95,99]

})
```

```python
df2
```

```python
pd.merge(df1,df2) ## defaultly it take common row as reference
```

### 1.Inner Merge (Default)

Returns only matching rows

```python
df1 = pd.DataFrame({
    "ID": [1,2,3],
    "Name": ["A","B","C"]
})

df2 = pd.DataFrame({
    "ID": [2,3,4],
    "Marks": [80,90,95]
})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, on="ID", how="inner") # give common rows
```

### 2.Left Merge

Returns all rows from the left DataFrame

```python
pd.merge(df1,df2,on='ID',how='left')
```

### 3.Right Merge

Returns all rows from the right DataFrame

```python
pd.merge(df1, df2, on="ID", how="right")
```

###4. Outer Merge

Returns all rows from both DataFrames.

```python
pd.merge(df1, df2, on="ID", how="outer")
```

## join

Use join() when combining DataFrames based on their index

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Harish", "Rahul", "Ravi"]
})

df2 = pd.DataFrame({
    "Marks": [85,90,95]
})
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

## concat

Use concat() to stack DataFrames either vertically (rows) or horizontally (columns)

### vertical conactination

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Harish", "Rahul"]
},index=[1,2])

df2 = pd.DataFrame({
     "Marks": [85,90]
},index=[1,2])
```

```python
pd.concat([df1, df2])
```

### reset index

```python
pd.concat([df1, df2], ignore_index=True)
```

## horizontal concatination

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Harish", "Rahul"]
})

df2 = pd.DataFrame({
    "Marks": [85,90]
})
```

```python
pd.concat([df1, df2], axis=1)
```
