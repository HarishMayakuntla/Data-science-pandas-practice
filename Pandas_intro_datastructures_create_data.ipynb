## What is a Series?

A Series is a one-dimensional data structure.
 * like list ,tuple ,set ,dict so we can use the single colume

 * series has only one dimenstion like the  1D array

* every value in series has index

 ### Syntax :

     pd.Series([elements])

```python
# series follwed by the index

price=pd.Series([20,50,60,40,80,900])
price
```

```python
# here left side is index   0   1   2  3  4   5
# and right sside is values 20  50  60 40 80  900
```

```python
price[0]  # here index works
```

```python
price[4]
```

### create custom index

Instead of default indexes (0, 1, 2...), you can define your own

```python
marks=pd.Series(
    [50,80,60],

    index=['maths','Telugu','social']

)

marks
```

```python
marks['maths']
```

### create series in dictionary

```python
student={
    'harish':70.0,
    'coures':'data science',
    'age':21,
    'fee':'none'
}

s=pd.Series(student)

s
```

```python
s['harish']
```

### create Series in  differnt data type

```python
## NUMBERS
pd.Series([20,30,50,60,5])
```

```python
## Strings
pd.Series(['mango','apple','orange'])
```

```python
## boolean

pd.Series([True,False,True,False,True])
```

```python
## float or decimals
pd.Series([2.5,6.25,600.865,873.5421])
```

# What is a DataFrame?

A DataFrame is a two-dimensional table
* it is also follwed by index

### Syntax :

         pd.DataFrame(data)

### create the DataFrame

```python
data={

      'name':['harish','ramesh','somesh'],
      'age':[21,22,22],
      'course':['datascience','python fullstack','data analytics']
}

df=pd.DataFrame(data)

df
```

```python
## above   index 0 1 2 and it represents the rows
## and the name , age ,course are columes
## the remaing is the data
```

```python
print(df["name"])
```

```python
df[['name','age','course']]
```

```python
df.ndim   ## dimenstons
```

```python
df.shape
```

```python
df.columns
```

```python
df.index
```

```python
df.dtypes
```

### From a List of Lists

```python
data = [
    ["Harish",22],
    ["Rahul",23],
    ["Ravi",21]
]

df = pd.DataFrame(data, columns=["Name","Age"])

df
```

### From a List of Dictionaries

```python
data = [
    {"Name":"Harish","Age":22},
    {"Name":"Rahul","Age":23},
    {"Name":"Ravi","Age":21}
]

df=pd.DataFrame(data)

df
```
