# pandas using String Operations, Apply, Map

## String Operations in Pandas

String operations are used when a DataFrame column contains text/string data

* we use .str before string methods

str.upper()
str.lower()
str.title()
str.len()
str.contains()
str.startswith()
str.endswith()
str.replace()
str.strip()

```python
## create the dataframe
import pandas as pd

df = pd.DataFrame({
    "Name": ["harish", "RAHUL", "sneha", "Kiran"],
    "City": ["hyderabad", "Delhi", "mumbai", "Bangalore"]
})

print(df)
```

```python
df['Name'].str.upper()
```

```python
df['Name'].str.lower()
```

```python
df["City"].str.title()
```

```python
df["Name_Length"] = df["Name"].str.len()

df
```

```python
df[df["Name"].str.contains("A")]
```

```python
df[df["Name"].str.startswith("h")]
```

```python
df[df["Name"].str.endswith("L")]
```

```python
df["City"].str.replace("Delhi", "New Delhi")
```

```python
df["Name"].str.strip()
```

## Apply()


apply() is used to apply a function to values in a Series or across a DataFrame

```python
df = pd.DataFrame({
    "Name": ["Harish", "Rahul", "Sneha"],
    "Marks": [85, 65, 92]
})
```

```python
# suppose i need to add the 5 marks to the marks

def add_marks(x):
    return x + 5

df['new_marks']=df['Marks'].apply(add_marks)
df
```

## Map()

map() is commonly used with a Series to transform or replace its values

```python
df = pd.DataFrame({
    "Name": ["Harish", "Rahul", "Sneha"],
    "Gender": ["M", "M", "F"]
})
df
```

```python
df['Gender']=df['Gender'].map({'M':'Male','F':'Female'})

df
```
