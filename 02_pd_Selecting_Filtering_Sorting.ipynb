# Selecting, Filtering, Sorting

## Create a DataFrame

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Priya", "Anil", "Sneha"],
    "Age": [22, 25, 21, 24, 23],
    "Marks": [85, 72, 91, 65, 78],
    "City": ["Kadiri", "Hyderabad", "Bangalore", "Kadiri", "Chennai"]
}

df = pd.DataFrame(data)

print(df)
```

# selecting Data

Selecting means getting particular columns or rows from a DataFrame



###  Select single column

* Syntax :
     
      df['column']

```python
## select one column

df['Name']
```

### Select multiple columns

   * Syntax :
     
      df [['column','column']]

```python

## slecting multiple columns

df[['Name','Age']] # multiple column
```

###  Selecting Rows — iloc

iloc means select using integer position

```python
df.iloc[0:3,0:3]
```

```python
df.iloc[2]
```

```python
df.iloc[0,[1,2]]
```

### Selecting Using -loc

loc selects using labels

it depends on integers

```python
df.loc[0]
```

```python
df.loc[0, "Name"]
```

```python
df.loc[0:2,['Name','Marks']]
```

```python
df.index = ["A", "B", "C", "D",'E']
df
```

```python
df.loc['B']
```

### difference between iloc  vs loc


| Feature           | `loc`               | `iloc`                   |
| ----------------- | ------------------- | ------------------------ |
| Meaning           | Label location      | Integer location         |
| Uses              | Labels              | Positions                |
| Row               | `df.loc["A"]`       | `df.iloc[0]`             |
| Column            | `df.loc[:, "Name"]` | `df.iloc[:, 0]`          |
| Multiple rows     | `df.loc["A":"C"]`   | `df.iloc[0:3]`           |
| Slice ending      | Included            | Excluded                 |
| Boolean filtering |                     | Usually not the main use |


# Filtering

Filtering means getting rows that satisfy a condition


### Filtering with <,>

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Priya", "Anil", "Sneha"],
    "Age": [22, 25, 21, 24, 23],
    "Marks": [85, 72, 91, 65, 78],
    "City": ["Kadiri", "Hyderabad", "Bangalore", "Kadiri", "Chennai"]
}

df = pd.DataFrame(data)

print(df)
```

```python
## i need check whose marks grater than 80

df[df['Marks']>80]
```

```python
df['Marks']>80  # here the otput as boolean type
```

```python
## i check age less than 21

df[df['Age']<22]
```

### Filter with == , <= ,>= ,!=

```python
df[df['City']=='Kadiri']
```

```python
df[df['Age']<=22]
```

```python
df[df['Marks']>=65]
```

```python
df[df['Name']!='Harish']
```

### Multiple Comditions

AND ----- >  &

OR ------->  |

```python
## we need check mutiple conditions at single time
## if the both condition must be true

df[(df['Marks']<85)&(df['Age']>23)]
```

```python
## in OR if only on condition is True
# it will give both condtion satisfys

df[(df['Marks']<85)|(df['Age']>23)]
```

### Filtering with isin()  , between ()

```python
## it better thanmultiple OR condtions

df[df["City"].isin(["Kadiri", "Chennai"])]
```

```python
## check in between values

df[df["Marks"].between(70,85)]
```

# Sorting

Sorting means arranging your data

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Priya", "Anil", "Sneha"],
    "Age": [22, 25, 21, 24, 23],
    "Marks": [85, 72, 91, 65, 78],
    "City": ["Kadiri", "Hyderabad", "Bangalore", "Kadiri", "Chennai"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.sort_values('Marks')
```

```python
df.sort_values('Marks',ascending=False)
```

```python
df.sort_values(['City','Marks'])
```

```python
df.sort_index(ascending=False)
```

```python
df.sort_index()
```

```python
df[df["Marks"] > 75][["Name", "Marks"]].sort_values(
    "Marks",
    ascending=False
)
```
