## Real Example

```python
import pandas as pd

data = {
    "Name": ["Harish", "Rahul", "Ravi"],
    "Age": [22, 21, 23],
    "Marks": [85, 78, 92]
}

dframe = pd.DataFrame(data)

dframe
```

```python
dframe.to_csv("student.csv",index=False)
```

```python
dframe
```

```python
dframe.to_excel("students.xlsx", index=False)
```

```python
dframe
```

```python
dframe.to_json("students.json", orient="records", indent=4)
```
